## Abstract

Professional tennis is played on three main surfaces: Hard, Clay, and Grass. Each offering different speeds and bounce heights. Our project aimed to see if these physical differences made certain surfaces more unpredictable, leading to more **upsets** where lower-ranked players defeat the favorites. We hypothesized that specialized surfaces like Clay or Grass would show higher upset rates than the more standard Hard courts.

To test this, we analyzed over 60,000 ATP matches, cleaning the data to remove missing values and focusing only on the primary court types. We created a variable called is_upset to track when an underdog won and compared these rates across all three surfaces. This allowed us to see if the environment truly gave lower-ranked players a statistical advantage.

Ultimately, our analysis showed that court surface does not have a significant impact on the likelihood of an upset. The upset rates across Hard, Clay, and Grass courts were remarkably similar, suggesting that professional rankings are a consistent predictor of success regardless of the playing conditions. While surfaces change how the game looks, they do not seem to change the fundamental probability of the better player winning the match.

## Research Question

**How does court surface (Hard, Clay, Grass) affect (1) upset probability and (2) the predictive power of player rankings in professional tennis matches?**

Specifically, we will examine:

- Whether the probability of an upset (lower-ranked player defeating higher-ranked player) differs significantly across surfaces.

- Whether ranking difference predicts match outcomes equally well on all surfaces.

**Key variables** include:

- Surface (categorical)

- Rank_1, Rank_2 (used to compute ranking difference)

- Winner (binary outcome)

- Odds (used as a secondary comparison baseline)

**Controls**
* **Tournament Type** (Grand Slams, Masters 1000, ATP 500/250) this will help isolate whether upsets are caused by court surface itself or differing levels of pressure and player motivation.

* **Match Format** (Best of 3 vs Best of 5) because Grand Slams hold best-of-5 matches and standard ATP matches are best-of-3 we will be using "best of" as a control variable, accounting for longer matches typically reducing the probability of an upsets by allowing a higher level player more time to recover from a slow start.

This project combines statistical inference (testing differences in upset rates across surfaces) with a prediction task (logistic regression models estimating match outcomes from ranking differences, with and without surface information). Model performance metrics such as accuracy and AUC will be used to quantify how much predictive power surface adds.


## Background and Prior Work

Professional tennis is played across multiple court surfaces—primarily hard, clay, and grass—and these surfaces materially change how matches unfold. The International Tennis Federation (ITF) formalizes these differences through its *Court Pace Rating* system, which classifies surfaces based on measured ball speed and bounce characteristics. These classifications reflect meaningful physical differences in gameplay, such as rally length and serve effectiveness, and provide an official basis for treating surface as a structural variable in match outcomes.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)

Sports science research supports the idea that surface type alters match dynamics. A recent large-scale match analysis study found statistically significant differences in rally structure, point duration, and serve impact across clay, grass, and hard courts, demonstrating that player performance profiles shift depending on surface conditions.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) These findings suggest that competitive balance may vary by surface, motivating investigation into whether certain environments produce greater outcome variability.

From a predictive modeling perspective, Klaassen and Magnus introduced probabilistic approaches for forecasting tennis match winners, showing that match outcomes can be effectively modeled using player performance indicators.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Their work provides a foundation for treating tennis outcomes as a statistical inference and prediction task, although it does not explicitly focus on surface-specific effects.

More directly related to our project goals, Kovalchik evaluated multiple tennis prediction methods and demonstrated that model accuracy depends strongly on how player strength is represented, highlighting limitations of relying on a single global ranking signal.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) This motivates examining whether ranking differences carry equal predictive meaning across different surfaces, rather than assuming rankings summarize player ability uniformly.

Applied analytics platforms have operationalized this idea through surface-aware rating systems. Tennis Abstract maintains separate Elo ratings for hard, clay, and grass courts and reports improved forecasting performance compared to a single universal rating.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) Additionally, Tennis Abstract has shown that match predictability varies by surface, with some surfaces consistently producing higher variance outcomes.<a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6) These findings align closely with our project’s focus on comparing upset rates and ranking predictability across surfaces.

Building on this prior work, our project integrates descriptive statistics and interpretable modeling to examine (1) whether upset probabilities differ by surface and (2) whether player ranking differences predict match outcomes equally well across environments. Unlike many existing analyses that emphasize either descriptive trends or pure prediction, we aim to combine statistical testing with model-based evaluation to better understand how surface mediates competitive structure in professional tennis.

## Hypothesis


We hypothesize that:

1. Upset rates will be significantly higher on clay courts than on hard or grass courts.

2. Ranking difference will be less predictive of match outcomes on clay compared to hard and grass surfaces.

Our reasoning is that clay courts slow down play and reduce the advantage of powerful serves, allowing lower-ranked players to remain competitive through longer rallies and defensive consistency. This should increase variance in outcomes and weaken the predictive signal of rankings. Conversely, grass and hard courts are expected to amplify skill and power differences, making rankings more informative predictors.

Rankings are unitless but serve as a proxy for relative skill level based on accumulated performance points over the previous 52 weeks. Ranking differences are commonly used in sports analytics as a simple measure of expected competitive imbalance; however, because rankings are ordinal rather than continuous measurements of skill, a gap between Rank 1 and Rank 10 may not reflect the same performance difference as Rank 100 versus Rank 110.

Comparing match outcomes against betting odds allows us to quantify how surprising a result was and provides a useful baseline against which statistical models can be evaluated.

#### Feature Engineering
We create a `rank_diff` feature to quantify the competitive gap between players using their ATP rankings. Since lower ranking numbers indicate stronger players, the absolute difference between `Rank_1` and `Rank_2` provides a simple measure of expected skill disparity. This variable will be used later to analyze upset probability and to evaluate how strongly rankings predict match outcomes across different court surfaces.

```python
df["rank_diff"] = (df["Rank_1"] - df["Rank_2"]).abs()
```

We then create an `is_upset` feature to identify matches where the lower-ranked player defeats the higher-ranked player. Since smaller ranking numbers represent stronger players, we first determine which player is the pre-match favorite using the ranking comparison (`Rank_1 < Rank_2`). 

The `Winner` column is inspected to confirm that it stores player names rather than numeric codes. If the encoding is valid, we compute `is_upset` directly by checking whether the non-favorite player won the match. This variable will be used later to measure upset probability across different court surfaces and evaluate how predictive rankings are under varying playing conditions.


```python
# Inspect Winner encoding before computing is_upset.
if all(c in df.columns for c in ["Winner", "Player_1", "Player_2"]):
    winner_matches_names = ((df["Winner"] == df["Player_1"]) | (df["Winner"] == df["Player_2"])).mean()
    print("\nWinner matches Player_1/Player_2 name rate:", winner_matches_names)

    if winner_matches_names > 0.5:
        favorite_is_p1 = df["Rank_1"] < df["Rank_2"]
        df["is_upset"] = 1 - np.where(
            favorite_is_p1,
            (df["Winner"] == df["Player_1"]).astype(int),
            (df["Winner"] == df["Player_2"]).astype(int),
        )
    else:
        print("Winner not encoded as player names; skipping favorite_won/is_upset for now.")
```

### Analysis 1: Do upset rates differ significantly by surface?

```python
## YOUR CODE HERE
## FEEL FREE TO ADD MULTIPLE CELLS PER SECTION
import scipy.stats as stats
import pandas as pd

contingency = pd.crosstab(df['Surface'], df['is_upset'])

chi2, p, dof, expected = stats.chi2_contingency(contingency)

n = contingency.to_numpy().sum()
min_dim = min(contingency.shape) - 1
cramers_v = np.sqrt(chi2 / (n * min_dim))

surface_summary = (
    df.groupby("Surface")["is_upset"]
      .agg(["mean", "sum", "count"])
      .rename(columns={"mean": "upset_rate", "sum": "num_upsets", "count": "num_matches"})
      .sort_values("upset_rate", ascending=False)
)

print("Upset summary by surface:")
display(surface_summary)

print(f"Chi-square statistic: {chi2:.4f}")
print(f"p-value: {p:.4f}")
print(f"Degrees of freedom: {dof}")
print(f"Cramer's V: {cramers_v:.4f}")
```

#### Results

We examined upset rates across surfaces and conducted a chi-square test of independence to determine whether upset probability differs by court type.

Observed upset rates were:
- Clay: 34.91% (5,554 / 15,909 matches)
- Grass: 34.48% (1,958 / 5,679 matches)
- Hard: 33.84% (9,760 / 28,843 matches)

The chi-square test yielded:
- X^2 = 5.39
- p-value = 0.0676

#### Conclusion

Since the p-value (0.0676) is greater than the standard significance threshold of 0.05, we fail to reject the null hypothesis. This indicates that there is no statistically significant evidence that upset rates differ across surfaces.

Although clay courts exhibit the highest upset rate and hard courts the lowest, the differences are small (approximately 1 percentage point). This is further supported by the very low effect size (Cramér’s V = 0.0103), indicating that surface has a negligible practical impact on upset likelihood.

Overall, while there is a slight trend toward higher upset rates on clay, the evidence suggests that court surface alone does not meaningfully influence the probability of an upset in professional matches.

### Analysis 2: Does surface affect how well rankings predict match outcomes?


```python
## YOUR CODE HERE
## FEEL FREE TO ADD MULTIPLE CELLS PER SECTION
import statsmodels.formula.api as smf
from scipy.stats import chi2

# Main-effects model
model_main = smf.logit('is_upset ~ abs_rank_diff + C(Surface)', data=df).fit()

# Interaction model
model_interaction = smf.logit('is_upset ~ abs_rank_diff * C(Surface)', data=df).fit()

print("=== Main-effects model ===")
print(model_main.summary())

print("\n=== Interaction model ===")
print(model_interaction.summary())

# Likelihood-ratio test: does interaction significantly improve fit?
lr_stat = 2 * (model_interaction.llf - model_main.llf)
df_diff = model_interaction.df_model - model_main.df_model
lr_p = chi2.sf(lr_stat, df_diff)

print("\nLikelihood-ratio test comparing interaction vs main-effects model:")
print(f"LR statistic: {lr_stat:.4f}")
print(f"df difference: {df_diff}")
print(f"p-value: {lr_p:.4f}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, accuracy_score
from sklearn.model_selection import train_test_split

surface_results = []

for surface in ["Hard", "Clay", "Grass"]:
    sub = df[df["Surface"] == surface].copy()

    X = sub[["abs_rank_diff"]]
    y = sub["is_upset"]

    X_train, X_test, y_train, y_test = train_test_split(
        X, y,
        test_size=0.3,
        random_state=42,
        stratify=y
    )

    clf = LogisticRegression(max_iter=1000)
    clf.fit(X_train, y_train)

    y_prob = clf.predict_proba(X_test)[:, 1]
    y_pred = clf.predict(X_test)

    surface_results.append({
        "Surface": surface,
        "n_matches": len(sub),
        "AUC": roc_auc_score(y_test, y_prob),
        "Accuracy": accuracy_score(y_test, y_pred),
        "RankGap_Coefficient": clf.coef_[0][0]
    })

surface_perf = pd.DataFrame(surface_results).sort_values("Surface")
display(surface_perf)
```

```python
surface_perf_plot = surface_perf.set_index("Surface")[["AUC", "Accuracy"]]
surface_perf_plot.plot(kind="bar", figsize=(8,5))

plt.title("Predictive Performance of Ranking Gap by Surface")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.show()
```

```python
df["favorite_wins"] = 1 - df["is_upset"]
df["rank_bin"] = pd.cut(df["abs_rank_diff"], bins=20)

rank_surface_pred = (
    df.groupby(["Surface", "rank_bin"])["favorite_wins"]
      .mean()
      .reset_index()
)

plt.figure(figsize=(10, 6))
for surface in ["Hard", "Clay", "Grass"]:
    sub = rank_surface_pred[rank_surface_pred["Surface"] == surface]
    plt.plot(range(len(sub)), sub["favorite_wins"], marker="o", label=surface)

plt.title("Favorite Win Rate vs Ranking Difference by Surface")
plt.xlabel("Ranking Difference Bin")
plt.ylabel("Favorite Win Rate")
plt.legend()
plt.show()
```

```python
# Keep only rows with usable odds
df_odds = df.dropna(subset=["Odd_1", "Odd_2"]).copy()

# Create favorite_odds as the odds of the higher-ranked player
df_odds["favorite_is_p1"] = df_odds["Rank_1"] < df_odds["Rank_2"]
df_odds["favorite_odds"] = np.where(
    df_odds["favorite_is_p1"],
    df_odds["Odd_1"],
    df_odds["Odd_2"]
)

# Target
y = df_odds["is_upset"]

# Model 1: ranking gap only
X_rank = df_odds[["abs_rank_diff"]]

# Model 2: betting odds only
X_odds = df_odds[["favorite_odds"]]

Xr_train, Xr_test, yr_train, yr_test = train_test_split(
    X_rank, y, test_size=0.3, random_state=42, stratify=y
)

Xo_train, Xo_test, yo_train, yo_test = train_test_split(
    X_odds, y, test_size=0.3, random_state=42, stratify=y
)

rank_model = LogisticRegression(max_iter=1000)
odds_model = LogisticRegression(max_iter=1000)

rank_model.fit(Xr_train, yr_train)
odds_model.fit(Xo_train, yo_train)

rank_prob = rank_model.predict_proba(Xr_test)[:, 1]
odds_prob = odds_model.predict_proba(Xo_test)[:, 1]

rank_pred = rank_model.predict(Xr_test)
odds_pred = odds_model.predict(Xo_test)

baseline_compare = pd.DataFrame({
    "Model": ["Ranking Gap Only", "Favorite Odds Only"],
    "AUC": [
        roc_auc_score(yr_test, rank_prob),
        roc_auc_score(yo_test, odds_prob)
    ],
    "Accuracy": [
        accuracy_score(yr_test, rank_pred),
        accuracy_score(yo_test, odds_pred)
    ]
})

display(baseline_compare)
```
